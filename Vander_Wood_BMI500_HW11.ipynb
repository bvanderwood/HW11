## Agent Based Modeiling of Pandemic Spread

### Imports and Setup

```python
import modelclasses
import simulation
```

```python
#Set up a bunch of configuration parameters
possible_fraction_infected = [0.05, 0.1, 0.2]
possible_ps = [0.1, 0.3, 0.7, 1]
possible_qs = [0.001, 0.02, 0.1]

params = []

for p in possible_ps:
    for q in possible_qs:
        for fraction_infected in possible_fraction_infected:
            params.append(modelclasses.SimulationConfig(
                grid_size=50, 
                agent_count=150,
                steps=300,
                social_distancing=False,
                fraction_infected=fraction_infected,
                p=p,
                q=q
            ))
```

### Try a bunch of different infection parameters, no social distancing

```python
# Run simulations

for param in params:
    print("-----------------------------")
    print(f"Running simulation with p={param.p}, q={param.q}, fraction_infected={param.fraction_infected}")
    #Make a grid
    grid = simulation.makeGrid(param)
    #Make agents
    agents = simulation.makeAgents(param)
    #Run simulation
    history = simulation.runSimulation(param)
    #View results
    simulation.collateResults(history)
    print("-----------------------------")
```

### Now add social distancing - we'll try it on the worst case we tested

```python
# Take the most dramatic parameter set and apply social distancing
param = modelclasses.SimulationConfig(
    grid_size=75, 
    agent_count=200,
    steps=400,
    social_distancing=True,
    fraction_infected=0.2,
    p=1,
    q=0.001
)
print("-----------------------------")
print(f"Running simulation with SOCIAL DISTANCING p={param.p}, q={param.q}, fraction_infected={param.fraction_infected}")
#Make a grid
grid = simulation.makeGrid(param)
#Make agents
agents = simulation.makeAgents(param)
#Run simulation
history = simulation.runSimulation(param)
#View results
simulation.collateResults(history)
print("-----------------------------")
```

### Let's try more dramatic social distancing...

## Expanded social distancing testing

```python
# Set up a bunch of random social distancing tests
possible_levy_probs = [0.5, 0.01]
possible_fraction_infected = [0.05, 0.2, 0.4]
possible_ps = [0.5, 0.65, 1]
possible_qs = [0.001, 0.02, 0.05]
params = []

for levies in possible_levy_probs:
    for p in possible_ps:
        for q in possible_qs:
            for fraction_infected in possible_fraction_infected:
                params.append(modelclasses.SimulationConfig(
                    grid_size=50, 
                    agent_count=100,
                    steps=300,
                    social_distancing=False,
                    fraction_infected=fraction_infected,
                    levy_prob_distancing=levies,
                    p=p,
                    q=q
                ))
                params.append(modelclasses.SimulationConfig(
                    grid_size=50, 
                    agent_count=100,
                    steps=300,
                    social_distancing=True,
                    fraction_infected=fraction_infected,
                    levy_prob_distancing=levies,
                    p=p,
                    q=q
                ))

# Run simulations
for param in params:
    if param.social_distancing:
        print("-----------------------------")
        print(f"Running simulation with SOCIAL DISTANCING p={param.p}, q={param.q}, fraction_infected={param.fraction_infected}, move distancing prob={param.levy_prob_distancing}")
    else:
        print("-----------------------------")
        print(f"Running simulation WITHOUT SOCIAL DISTANCING p={param.p}, q={param.q}, fraction_infected={param.fraction_infected}, move distancing prob={param.levy_prob_distancing}")
    #Make a grid
    grid = simulation.makeGrid(param)
    #Make agents
    agents = simulation.makeAgents(param)
    #Run simulation
    history = simulation.runSimulation(param)
    #View results
    simulation.collateResults(history)
    print("-----------------------------")
```
